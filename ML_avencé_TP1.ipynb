```python
import numpy as np
from keras.datasets import mnist
from keras.utils import to_categorical
from keras import backend as K

K.set_image_data_format('channels_first')

def get_and_prepare_data_mnist():
    
    (X_train, y_train), (X_test, y_test) = mnist.load_data()
    
    X_train = X_train.reshape(X_train.shape[0], 1, 28, 28).astype('float32')
    X_test = X_test.reshape(X_test.shape[0], 1, 28, 28).astype('float32')
    
    X_train = X_train / 255
    X_test = X_test / 255
    
    y_train = to_categorical(y_train)
    y_test = to_categorical(y_test)
    
    num_classes = y_test.shape[1]
    
    return (X_train, y_train), (X_test, y_test), num_classes
```

```python
(X_train, y_train), (X_test, y_test), num_classes = get_and_prepare_data_mnist()

print("Shape X_train :", X_train.shape)
print("Shape y_train :", y_train.shape)
```

### Construction du Small CNN

```python
from keras.models import Sequential
from keras.layers import Conv2D, Flatten, Dense

# Création du modèle
model = Sequential()

# 1ère couche convolution
model.add(Conv2D(64, (3,3), input_shape=(1,28,28), activation='relu'))

# 2ème couche convolution
model.add(Conv2D(32, (3,3), activation='relu'))

# Flatten
model.add(Flatten())

# Couche Dense finale
model.add(Dense(num_classes, activation='softmax'))

# Compilation
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

>Dans cette étape, nous avons construit un réseau de neurones convolutionnel composé de :

>Deux couches Conv2D pour extraire les caractéristiques des images.

>Une couche Flatten pour transformer les matrices en vecteur.

>Une couche Dense avec Softmax pour prédire les probabilités des 10 chiffres.

>Ensuite, nous avons compilé le modèle en utilisant :

>>categorical_crossentropy comme fonction de perte.

>>adam comme optimiseur.

>>accuracy comme métrique.

### Entraînement du modèle

```python
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=10,
    batch_size=200
)
```

>Nous avons entraîné le modèle pendant 10 epochs.

>X_train, y_train : données d’apprentissage.

>validation_data : permet de tester le modèle à chaque epoch.

>batch_size=200 : nombre d’images traitées avant mise à jour des poids.

### Évaluation du modèle

```python
scores = model.evaluate(X_test, y_test)

print("Accuracy :", scores[1])
print("Error rate :", 1 - scores[1])
```

>Nous avons évalué le modèle sur les données de test.

>scores[1] représente l’accuracy.

>Le taux d’erreur = 1 − accuracy.

>Normalement tu dois obtenir :

>Accuracy ≈ 0.98
>Erreur ≈ 0.01

```python
# ==============================
# PARTIE E2 - Évaluation du modèle
# ==============================

# Fonction pour évaluer la performance du modèle
def print_model_error_rate(model, X_test, y_test):
    # Évalue le modèle sur les données de test
    scores = model.evaluate(X_test, y_test, verbose=0)
    
    # Affiche la précision
    print("Model score : %.2f%%" % (scores[1]*100))
    
    # Affiche le taux d'erreur
    print("Model error rate : %.2f%%" % (100 - scores[1]*100))

# Appel de la fonction pour afficher la précision et l'erreur
print_model_error_rate(model, X_test, y_test)
```
