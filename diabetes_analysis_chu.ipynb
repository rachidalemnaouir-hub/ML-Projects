>> Suppression des lignes avec valeurs manquantes

```python
# Supprimer les lignes qui contiennent au moins une valeur manquante
df_sans_na = df.dropna()
```


 
 
 >>Imputation des valeurs manquantes
> 
> 

```python
colonnes_numeriques = df.select_dtypes(include=['float64', 'int64']).columns

for col in colonnes_numeriques:
    if df[col].isnull().sum() > 0:
        moyenne = df[col].mean()
        df[col].fillna(moyenne, inplace=True)
        print(f"Valeurs manquantes dans {col} remplies avec la moyenne : {moyenne}")
```

```python
colonnes_categorielles = df.select_dtypes(include=['object']).columns

for col in colonnes_categorielles:
    if df[col].isnull().sum() > 0:
        mode_val = df[col].mode()[0]
        df[col].fillna(mode_val, inplace=True)
        print(f"Valeurs manquantes dans {col} remplies avec : {mode_val}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import numpy as np



sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 5.1 Graphiques personnalisés
sns.histplot(data=df, x='age', bins=30, kde=True, color='skyblue')
plt.title("Distribution des âges")
plt.tight_layout()
plt.show()

sns.countplot(data=df, x='gender', palette='Set2')
plt.title("Répartition des patients par genre")
plt.tight_layout()
plt.show()

sns.countplot(data=df, x='hypertension', hue='diabetes', palette='husl')
plt.title("Hypertension en fonction du diabète")
plt.tight_layout()
plt.show()

sns.scatterplot(data=df, x='age', y='blood_glucose_level', hue='diabetes', palette='coolwarm')
plt.title("Relation entre l'âge et le taux de glucose")
plt.tight_layout()
plt.show()

# 5.2 Mise en évidence des patterns ou anomalies
plt.figure(figsize=(10, 8))
corr_matrix = df.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='YlGnBu')
plt.title("Carte de chaleur des corrélations")
plt.tight_layout()
plt.show()

fig = px.scatter(df, x="age", y="blood_glucose_level", color="diabetes", 
                 title="Clusters potentiels: Âge vs Glucose",
                 labels={"age": "Âge", "blood_glucose_level": "Taux de glucose"})
fig.show()

for col in ['bmi', 'HbA1c_level', 'blood_glucose_level']:
    sns.boxplot(data=df, x=col, color='lightgreen')
    plt.title(f"Boxplot : {col} (détection des valeurs extrêmes)")
    plt.tight_layout()
    plt.show()
```
